## Wholly invented: exact pairing before scoring

The next cell uses two invented events. It checks scores against hand calculations, then changes one event identity while preserving the count. The scorer must reject the comparison. The generic timing checks validate declared clocks; they do not authenticate historical possession or reconstruct any private feature values.


```python
from dataclasses import replace
import math
from examples.fbs_contracts.matched_cohort import invented_comparison, score_matched, pool_additive

models = invented_comparison()
paired = score_matched(models)
assert math.isclose(paired[0]["metrics"]["brier"], (.2**2 + .3**2) / 2)
assert math.isclose(paired[0]["metrics"]["logloss"], -(math.log(.8)+math.log(.7))/2)
assert paired[0]["cohort_sha256"] == paired[1]["cohort_sha256"]
bad = invented_comparison()
bad["invented-challenger"][0] = replace(bad["invented-challenger"][0], event="other-invented-event")
try:
    score_matched(bad)
except ValueError as error:
    rejection = str(error)
else:
    raise AssertionError("Equal-count mismatch was admitted")
assert "exact event identities" in rejection
assert math.isclose(pool_additive([{"games":1,"metrics":{"rmse":2}}, {"games":3,"metrics":{"rmse":6}}], "rmse"), math.sqrt(28))
fixture_rows = "".join(f"<tr><td>{r['model']}</td><td>{r['games']}</td><td>{r['metrics']['brier']:.6f}</td><td>{r['metrics']['logloss']:.6f}</td></tr>" for r in paired)
display(HTML("<h3>Wholly invented two-event comparison</h3><table><thead><tr><th>Fixture</th><th>Games</th><th>Brier</th><th>Log loss</th></tr></thead><tbody>" + fixture_rows + "</tbody></table><p>Equal-count identity substitution: BLOCKED.</p>"))
print("No actual sports skill follows from these invented metrics. Weighted RMSE uses squared error, not a mean of RMSEs.")
```
